```python
from qiskit_dynamics.models import HamiltonianModel
from qiskit_dynamics.solvers import Solver
from qiskit_dynamics.signals import Signal
import numpy as np
from qiskit.quantum_info import Pauli
import matplotlib.pyplot as plt
from qiskit.quantum_info.states import Statevector
import qiskit.visualization as qv
from IPython.display import display, clear_output, update_display
import time
from mpl_toolkits.mplot3d import Axes3D
y0 = Statevector([1./np.sqrt(2), 1./np.sqrt(2)])
#y0 = Statevector([1, 0])
X = Pauli("X").to_matrix()
Y = Pauli("Y").to_matrix()
Z = Pauli("Z").to_matrix()
print(X)
print(Z)
def Rabi_impulse(t):
    #return np.sin(6.28*t)
    return 1
def f(x):
    return 1
Freq_0 = 100.0
Freq_E = 99.98 #9.98
Freq_Rabi = 1.0
vectorized_Rabi_impulse = np.vectorize(Rabi_impulse)
signal = Signal(envelope=vectorized_Rabi_impulse, carrier_freq = Freq_E, phase = 0)
#signal.draw(0, 6.28, 100)
#H_0 = 0
H_0 = 0.5 * 2 * np.pi * Freq_0 * Z
H_field = 2 * np.pi * Freq_Rabi * X
t_eval = np.linspace(0., 0.5, 1000)
hamiltonian = HamiltonianModel(static_operator=H_0, operators=H_field)
evolution = Solver( static_hamiltonian = H_0, hamiltonian_operators = [H_field])
result = evolution.solve(t_span=[0., .5], y0=y0,signals=[signal], t_eval=t_eval,     atol=3e-14,rtol=3e-14)
res = result
length = len(res.y)
bloch_x = np.zeros(len(res.y))
bloch_y = np.zeros(len(res.y))
bloch_z = np.zeros(len(res.y))
for i in range (len(res.y)):
    bloch_x[i] = res.y[i].expectation_value(X).real
    bloch_y[i] = res.y[i].expectation_value(Y).real
    bloch_z[i] = res.y[i].expectation_value(Z).real
   # bloch_x[i] = np.vdot(res.y[i],  X @ res.y[i]).real
   # bloch_y[i] = np.vdot(res.y[i], Y @ res.y[i]).real
   # bloch_z[i] = np.vdot(res.y[i], Z @ res.y[i]).real
center = np.array([0,0,0])
length = 1.2
#fig = plt.figure()
#ax = fig.add_subplot(111, projection='3d')
n = 100
xs = np.linspace(0, 1, n)
ys = np.sin(xs * 6 * np.pi)
zs = np.cos(xs * 6 * np.pi)
plt.style.use('_mpl-gallery')

plt.figure(2)
fig2, ax2d = plt.subplots()
plt.plot(t_eval, bloch_x, label = '$\\langle X \\rangle$')
plt.plot(t_eval, bloch_y, label = '$\\langle Y \\rangle$')
plt.plot(t_eval, bloch_z, label = '$\\langle Z \\rangle$')

plt.figure(1)
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.quiver(*center, length, 0, 0, color='r', linewidth=1.5)  # X
ax.quiver(*center, 0, length, 0, color='g', linewidth=1.5)  # Y
ax.quiver(*center, 0, 0, length, color='b', linewidth=1.5)  # Z
ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])
ax.set_xlim([-1, 1])
ax.set_ylim([-1, 1])
ax.set_zlim([-1, 1])
ax.text(length, 0, 0, 'X', color='r')
ax.text(0, length, 0, 'Y', color='g')
ax.text(0, 0, length, 'Z', color='b')
ax.set_box_aspect([1, 1, 1])
u, v = np.mgrid[0:2*np.pi:25j, 0:np.pi:25j]
x_sphere = np.cos(u)*np.sin(v)
y_sphere = np.sin(u)*np.sin(v)
z_sphere = np.cos(v)
ax.plot_surface(x_sphere, y_sphere, z_sphere, color =(0, 0.6, 1, 0.4))
point = ax.scatter(bloch_x[0], bloch_y[0], bloch_z[0], color='red', s=50)
quiv = ax.quiver(*center, bloch_x[0], bloch_y[0], bloch_z[0], color='red', linewidth=1.5)
print(res.y[1])
vect = display(qv.plot_bloch_vector([bloch_x[-1], bloch_y[-1], bloch_z[-1]],
                          f'Bloch vector at $t = {t_eval[-1]}$'), display_id = True)
for i in range(len(res.y)):
    if i % 1 == 0:
        #quiv.remove()
        #point = ax.scatter(bloch_x[i], bloch_y[i], bloch_z[i], color='red', s=50)
        #quiv = ax.quiver(*center, bloch_x[i], bloch_y[i], bloch_z[i], color='red', linewidth=1.5)
        #point._offsets3d = ([bloch_x[i]], [bloch_y[i]], [bloch_z[i]])
        #ax.scatter(bloch_x[i], bloch_y[i], bloch_z[i], color='red', s=50)
        #quiv._offsets3d = ([bloch_x[i]], [bloch_y[i]], [bloch_z[i]])
        #plt.pause(0.01)
        clear_output(wait=True)  # стираем предыдущий вывод
        update_display(qv.plot_bloch_vector([bloch_x[i], bloch_y[i], bloch_z[i]],
                                     f'Bloch vector at $t = {t_eval[-1]}$'), display_id=vect)
        time.sleep(0.5)
while True:
    plt.pause(0.01)
#bloch_params[0] = np.vdot(result[ya])
```
